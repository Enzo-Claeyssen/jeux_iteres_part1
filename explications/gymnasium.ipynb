# Qu'est-ce que Gymnasium ?

Gymnasium est une librairie qui permet la gestion des environnements en RL.

La majorité des librairies d'apprentissage par renforcement se basent sur les classes abstraites pour les environnements dans gymnasium.

# Comment créer un environnement ?

- Un environnement = une classe python
- La classe hérite de gymnasium.Env

- Il initialise l'environnement en le mettant dans un état initial

- Elle prend deux paramètres : seed et options qui valent None par défaut
- On commence par appeler la méthode reset() de la super classe sans oublier de faire passer la seed
- On réinitialise l'environnement en le mettant dans un état initial
- Renvoie l'état puis le dictionnaire d'info

# Qu'est-ce que gymnasium.spaces ?

Afin de décrire l'ensemble des action est l'ensemble des états (action_space et observation_space), on utilise l'outils spaces de gymnasium.

Il existe de multiples spaces différents selon l'ensemble à décrire :

```python
from gymnasium import spaces


"""
spaces pour une unique information
À utiliser si l'action / l'observation de l'agent n'utilise qu'une seule information
(Pour les observations : une position, une distance, une coordonnée etc...)
(Pour les actions : un choix [haut, bas, gauche ou droite ?], miser un certains nombre de jetons
"""

# Un ensemble fini de cardinal n
spaces.Discrete(n)
# OU
spaces.Box(low = 0, up = n - 1, dtype = int)

# Ces deux façons de coder un ensemble fini de carnidal n peuvent être interprétées différement par les algorithmes d'apprentissage
# Certains algorithmes sont compatibles avec certains spaces mais pas d'autres.
# Parfois cela peut changer la façon dont l'information est codée.
# Cela sera précisé dans les fiches des algorithmes concernés.




# Un interval [a, b]
spaces.Box(low = a, up = b)  # Pour des observations / actions continues




"""
spaces pour plusieurs informations
À utiliser si l'action / l'observation de l'agent utilise de multiples informations
(Pour les observations : position + état des cases adjacentes, plusieurs coordonnées)
(Pour les actions : si on contrôle plusieurs personnages, l'action effectuée par chacun d'entre eux)
"""


# Plusieurs informations chacune sur un ensemble discret
# Exemple : observer un entier dans [0, 1] ainsi qu'un entier dans [0, 3]
spaces.MultiDiscrete([2, 4])


# De multiples informations dans des ensembles potentiellement différents
# Exemple : observer pile ou face ainsi qu'un poids (un réel) dans [a, b]
spaces.Dict(
    {
        'pile_ou_face' : spaces.Discrete(2),
        'poids' : spaces.Box(low = a, up = b)
    }
)
```
